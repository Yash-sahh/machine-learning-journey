```python
# Real data

# Oscar Winner Movies
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years = [2008,2009,2010,2011,2012]
oscar_revenue = [1005,170,427,133,232] # in $M

# let we want to plot kitna revenue aaya in particular year
plt.plot(years,oscar_revenue) 

plt.title("Oscar Movies Revenue in ech Year") # to show title of the plot
plt.xlabel("Years") # to show what is in x-axis
plt.ylabel("Revenue {in $M}") # to show what is in y-axis
```

```python
# Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire","Avatar","Incepttion","Hugo","Lincoln"]
non_oscar_revenue = [378,2788,829,185,275] # in $M

# how plot multiple plots in same graph
# let see it by plotting oscar movies revenue and non oscar movies revenue by year in same graph

plt.plot(years,oscar_revenue, label = "Oscar Movies")  # if we are plotting multiple plot in same graph then we should pass label to distinguish between plots ki konsa plot kis ke liye bana hai
plt.plot(years,non_oscar_revenue, label = "Non-Oscar Movies")

plt.title("Oscar Movies vs Non-Oscar Movies Revenue")
plt.xlabel("Year")  
plt.ylabel("Revenue (in $M)")

plt.legend(loc = "upper center")  # we have created label to distinguish which plot is of which data by passing label in plt.plot ab iss label ko visible karane ke liye we have legend() which creates a legend box 
# also we can control the location of legend box via passing location as loc = 
# other wise simple plt.legend() kardo]
```

```python
# fromat string -> is is used to change the line of plot its color , line design etc
# in format string we have marker line color
# fmt = '[marker][line][color]'

plt.plot(X,Y,'d--r') # in this we have passed fromat string as 'd--r' where d= marker diamond , -- = line, r = red color
```

```python
# let we want to change only one thing not all marker line colour we want only color
# or we want ki apan sab alag alag kare not 
plt.plot(X,Y,color = 'r',linewidth = 3,linestyle = '-.')
plt.style.use("default")
plt.grid() # it shows the grid
```

## Plot Methods
1. plt.plot(x,y) -Line plot
2. plt.bar(x,height) -Vertical bar chart
3. plt.barh(y,width) -Horizontal bar chart
4. plt.scatter(x,y) - Scatter plot
5. plt.pie(data) - Pie chart
6. plt.hist(data,bins) - Histogram
7. plt.boxplot(data) -Box plot 

```python
#Bar Chart -> A bar plot ia a graph that displays categorical data using rectangular bars 
# vertical bar -> x- axis -> categories
#                 y- axis -> values
# horizontal bar -> x- axis -> values
#                 y- axis -> categories
movies = [
    "The Dark Knight -- 2",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years = [2008,2009,2010,2011,2012]
revenue = [1005,170,427,133,232]

plt.bar(years,revenue)
plt.title("Revenue of Oscar movies by year")
plt.xlabel("Years")
plt.ylabel("Revenue ($M)")

"""
if we want to print something on top of bar we can do it by 
plt.text(x-axis pos, y-axis pos, "text", alignment) 
"""
for i in range(len(years)):
    plt.text(years[i],revenue[i]+5,movies[i],ha = "center")# in y axis we did +5 taki thoda gap aa jaye and text thoda uper print ho

# if we want to increase the size of graph we can increase the limit of both y and x axis by using xlim and ylim
plt.ylim(0,max(revenue)+100) # similar for xlim minimum and maximum jitna ham rakhna chahte hai pass kar do
 
```

```python
# Comparing 2 sets of data with the help of bar chart
oscar_movies = [
    "The Dark Knight -- 2",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
oscar_revenue = [1005,170,427,133,232]

non_oscar_movies = [
    "Slumdog Millionaire",
    "Avatar",
    "Incepttion",
    "Hugo",
    "Lincoln"
]
non_oscar_revenue = [378,2788,829,185,275]

years = [2008,2009,2010,2011,2012]

# we want to compare revenue of oscar movies and non oscar movies year wise

plt.bar(years,oscar_revenue)
plt.bar(years,non_oscar_revenue)
plt.title("Oscar vs Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue($M)")
# now graph will be overlapping of these two graphs so we will have to change the width 
```

```python
# it can be handle 
#1) change the width from by default 0.8 to 0.4
import numpy as np
width = 0.4
x = np.arange(len(years))
plt.bar(x-width/2,oscar_revenue,width, label = "Oscar Movies")
plt.bar(x+width/2,non_oscar_revenue,width, label = "Non-Oscar Movies")
plt.title("Oscar vs Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue($M)")
plt.legend() 
# now x values are 0,1,2,.. we want to replace it with year
plt.xticks(x,years)  # we are not using year diractly while plotting because we have to handle the overlapping condition so we use x where x is an numpy array from 0 to len of years
```

```python
#horizontal bar chart
# category -> y axis
# value -> x axis
movies = [
    "The Dark Knight -- 2",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
revenue = [1005,170,427,133,232]
# lets print the revenue of every movie

plt.barh(movies,revenue)  # plt.barh(y-axis value, x-axis value) # opposite of bar plt.bar(x-axis value, y-axis value)
plt.xlabel("Revenue(M$)")
plt.ylabel("Movie names")
plt.title("Revenue by Movies")
```

```python
# Scatter plot -> it shows the relationship betweeen two variables by plotting point on a 2D plan
""" scatter plot are usefull when we want to see in our data
1) correlation
2) outlier detection
3) clusters
"""
people = ["Person A","Person B","Person C","Person D","Person E",
          "Person F","Person G","Person H","Person I","Person J"]
age = [22,25,30,35,40,45,50,55,60,65]
blood_pressure = [110,115,120,122,125,130,135,123,145,150]

# lets plot a scatter plot age vs blood pressure
plt.scatter(age,blood_pressure)
plt.title("Age vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()
# now we can get some insights like as age is increasing BP is increasing and there is one person that looking outside the common looking trend rest all are of increasing bp as age increases but this one is comaparitively lesser bp then his neighbour age peoples so it is an outlier
```

```python
colors = ["red","purple","yellow","blue","pink","red","purple","yellow","blue","pink"]
plt.scatter(age,blood_pressure,color = colors)
plt.title("Age vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()
```

```python
# bp>=135 red
# bp<135 green
colors = ["green" if x < 135 else "red" for x in blood_pressure ]
plt.scatter(age,blood_pressure,color = colors)
plt.title("Age vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()
```

```python
# lets plot jaha per bp jyada hai size jyada to sizes ke liye bp value ko hi pass kar do
plt.scatter(age,blood_pressure,s = blood_pressure) # for size we use s
plt.title("Age vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()
plt.show()
```

```python
# can also define marker
plt.scatter(age,blood_pressure,marker = '<') # for size we use s
plt.title("Age vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()
plt.show()
```

```python
#we can also define transparency of points mtlb points kitne visible honge and for it we define alpha value wich is 0 to 1 where 0 means no visible and 1 is completely visible and beech ki value translucent and by default alpha is 1 and we can set that alpha value
plt.scatter(age,blood_pressure,alpha = 0.5) # for size we use s
plt.title("Age vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()
plt.show()
```
