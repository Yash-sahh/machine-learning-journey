```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
```

```python
iris = load_iris()

X = pd.DataFrame(iris.data)
y = iris.target
```

```python
X.head()  # we have 4 features lets reduce it to 2 by using pca
```

```python
#.1) standardize data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# PCA
from sklearn.decomposition import PCA

pca = PCA(n_components=2)  # n_components means finally ham kitne features rakhna chahte hai

X_pca = pca.fit_transform(X_scaled)
```

```python
X_pca.shape
```

```python
# we can also see in which principal component kitna variance spread hua
print("explained variance ratio: ",pca.explained_variance_ratio_)
```

```python
# can also see eigon vectors
print(pca.components_)
```

```python
# visualize

plt.figure(figsize=(10,8))

plt.scatter(
    X_pca[:,0],
    X_pca[:,1]
)  # here we have plot original points that are extracted from pc1 and pc2
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA for iris dataset ")
```
