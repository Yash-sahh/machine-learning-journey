```python
#Transforming Data (Feature Engineering)
#1) apply() -> in this we write a logic to transform data in a function and then pass that function to apply to apply that transformation as apply(function)
#example  we want to create a new column for tax where we assign tax as if salary is >50k->20% and if<=50k -> 10%
df2 = df.copy()

df2["tax"] = df2["income"].apply(lambda x : "20%" if x>=50_000 else "10%")  # here we are passing an another function in apply function if there is any function where we pass function then that function is called higher order fuction
df2

```

```python
# 2 map() -> if we want to map existing values with another values
# for example let we want to map (or convert) Male -> M, Female -> F and Unknown to U

gender_map = {
    "Male":"M",
    "Female":"F",
    "Unknown":"U"
}
df2["gender"]=df2["gender"].map(gender_map) # here inside map we pass dictionary
df2
```

```python
#3 assign() -> can create a new column by actually assigning them some value
# let income of all is increased by 10%
df2 = df2.assign(new_income = df2["income"]*1.1)
df2
```

```python
# 4) replace(old,new) -> we can replace old value with new value
# let replace USA with US
df2["country"] = df2["country"].replace("USA","US")
df2
```

```python
#5) rename(columns = {old:new}) -> it changes column name
df2.columns = ["Id","Name","Age","Country","Gender","Income","Tax","New_Income"] # -> by this we can rename all column name

# but if we want to rename a particular column then
df2 = df2.rename(columns = {"Income":"Salary"})
df2
```

```python
# similarly if we want to rename row label we can do that also
df2.rename(index = {1:"first"})
```

```python
#6) sort_values -> we can sort the whole data on the basis of some quantitive column
# let suppose we want to sort data on the basis of Salary
df2["Salary"].sort_values() #-> it sorts "salary" column
df2.sort_values("Salary") #-> it sorts whole data on the basis of "Salary" column

sorted_df = df2.sort_values(["Salary","Age"]) # -> sorting will be on the basis of salary but if two incomes are same then it will look for their age and on the basis of that sorting will happen


# all above sorting is in the ascending order but if we want this sorting to be in descending order then we can 

df2.sort_values(["Salary","Age"],ascending=False) 
```

```python
# 7) sort_index() -> we can also sort on the basis index
# in sorted_df index are jumbled and we want whole data to be sorted on the basis of index values again 
sorted_df.sort_index()
```

```python
# 8) reset_index() ->  sort_index() function sorts the data on the basis of index to jo hamne pahle salary and age ke basis pe sort kiye the wo to gya but if we want ki data remain sorted but index values jo jumbled hui hai wo bas reset ho jaye
sorted_df.reset_index() # ab panda data loss nhi hone deta to ham iss function se index ko reset kar de rhe but yadi hame purane indexes ki jarurat padi to ? thats why pandas creates a new column index and usme purana index names preserved kar ke rakh de rha hai and hame index reset chahiye tha to usne naya label bhi create kar diya har ek row ke liye


# but if we do not want that original index to be preserved then we can
sorted_df.reset_index(drop=True)
```

```python
# 9) Rank() -> we can assign rank to our data from lowest to highest value
# let we want to create ranking on the basis of salary
sorted_df["Ranking"] = sorted_df["Salary"].rank()
sorted_df

# here rank 1 is lowest value but if we want it to be highest
sorted_df["Ranking"] = sorted_df["Salary"].rank(ascending=False)
sorted_df

# also if for example rank number 3 and 4 clash means here on the basis of salary both have same salary then both will be assigned rank = (3+4)/2 = 3.5 similar if 6 and 7 clashes rank of both will be (6+7)/2

# but if we want that rank shouldnt assign by this we want only integers to be as a rank value then we can do 
sorted_df["Ranking"] = sorted_df["Salary"].rank(method = "dense")
sorted_df  # now here if 3 and 4 clashesh then both will assign 3 and next will start from 4

# now if we want ki let say yeha par salary ke basis pe rank kar rhe to after sorting kuch esa hai esa hai ki 1 ,2 ,3 ,4 5, 6, ..  inme let say 4 and 5 are same to we want ranking esi ho 1,2,3,4,4,6.. then 
sorted_df["Ranking"] = sorted_df["Salary"].rank(method = "min")
sorted_df 

# and if we want it to be as 1,2,3,5,5,6.. then
sorted_df["Ranking"] = sorted_df["Salary"].rank(method = "max")
sorted_df 
```

```python
# let we want to group data on the basis of country
df.groupby("country")
# it grpoups our data and now we apply aggregation function to them such as sum/mean/max/min/count etc which performs that respective operation on that group
df.groupby("country")["income"].mean() #-> it groups data on the basis of country and we get mean of each countries income
df.groupby("country")["income"].min() # -> it gives maximum income of each country


#let we want to find mean income of genders -> gender ka income mean calculate karna hai to we will groupby it according to gender and then will apply mean on income
df.groupby("gender")["income"].mean()
```

```python
# agg() if we want more then one aggregation let assume we want to calculate min and max and mean then we will apply agg()

df.groupby("country")["income"].agg(["mean","min","max"])
df.groupby("country")["income"].aggregate(["mean","min","max"])   # both does same work

# here now columns name is mean min max and let we want to rename column
df.groupby("country")["income"].agg(avg_salary = "mean",min_salary = "min",max_salary="max")   

#now let we want ki ham har country ki income and age dono ka mean chahiye to ham ye bhi kar sakte hiai by using agg()
df.groupby("country").agg({
    "income":"mean",
    "age":"mean"
})

# now let say we want mean,min,max of income and age of every country
df.groupby("country").agg({
    "age":["mean","min","max"],
    "income":["mean","min","max"]
})

# now if we want to change column name
df.groupby("country").agg(
    max_salary = ("income","max"),
    avg_age = ("age","mean")
)
```

```python
df_customers = pd.DataFrame({
    "customer_id":[1,2,3,4],
    "name":["Adam","Bob","Charlie","Dave"]
})

df_orders = pd.DataFrame({
    "order_id":[101,102,103,104],
    "customer_id":[2,1,4,5],
    "amount":[250,120,300,180]
})
```

```python
df_customers
```

```python
df_orders
```

```python
# if we want to merge df_customer and df_orders
#1 inner join
pd.merge(df_customers,df_orders,on="customer_id") # on hai kiske basis pe merge kar rhe ek prakar se primary key and foreign key hai jo bhi customer id dono table me hai sirf wahi aayega after merge
```

```python
#2 Left join
pd.merge(df_customers,df_orders,on="customer_id",how="left") # left table ki sari values rakhenge and if right me vo nhi rha to use Nan se fil kardo
```

```python
#right join
pd.merge(df_customers,df_orders,on="customer_id",how="right") # right table ki sari value rakhenge or left ki wo values jo nhi hongi right ki ko rakhte hue use NaN se replace karenge
```

```python
# outer join
pd.merge(df_customers,df_orders,on="customer_id",how="outer") # dono hi table ki sari values layenge or jaha par bhi missing hai use NaN se replace karenge
```
