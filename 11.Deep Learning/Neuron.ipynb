```python
import torch
import torch.nn as nn
```

```python
# input 
torch.manual_seed(42) # every time we run a nuearal network code it is giving different value as output because every time weights and bias value is diffrent but if we do manual_seed in starting we got same output averytime it is like random state in ml
inputs = torch.tensor([1.0,2.0,3.0]) # tensors are multidimensional array which we can create inside pytorch and send data in float not integer 

```

```python
neuron = nn.Linear(in_features=3,out_features=1)  # linear activation function calculates weighted sum
```

```python
output = neuron(inputs)
```

```python
output
```

```python
# verify ... 
weights = neuron.weight # torch.tensor([-0.5663,  0.3475, -0.0782])
bias =    neuron.bias # torch.tensor([-0.4772])

output = weights @ inputs + bias  # we have verified by applying actual formula of weighted sum and @ is used to calculate dot product
print(output)
```
