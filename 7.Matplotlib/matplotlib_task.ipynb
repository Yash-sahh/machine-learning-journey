```python
# Task
days = ["Mon","Tue","Wed","Thu","Fri"]

cities = ["New York","London","Delhi","Tokyo"]

temperature = [
    [22,23,21,24,25], # New York
    [18,19,17,20,21], # London
    [30,32,31,33,34], # Delhi
    [25,26,24,27,28]  # Tokyo
]

# Compare temperature of cities over a week using subplots and loops
import matplotlib.pyplot as plt
fig,ax = plt.subplots(2,2)

ax[0][0].plot(days,temperature[0])
ax[0][0].set_title("New York")
ax[0][0].set_xlabel("days")
ax[0][0].set_ylabel("temp")

ax[0][1].plot(days,temperature[1])
ax[0][1].set_title("London")
ax[0][1].set_xlabel("days")
ax[0][1].set_ylabel("temp")

ax[1][0].plot(days,temperature[2])
ax[1][0].set_title("Delhi")
ax[1][0].set_xlabel("days")
ax[1][0].set_ylabel("temp")

ax[1][1].plot(days,temperature[3])
ax[1][1].set_title("Tokyo")
ax[1][1].set_xlabel("days")
ax[1][1].set_ylabel("temp")

fig.tight_layout()
# this was individual not using loop 
# we will plot using loop in next cell
```

```python
fig,ax = plt.subplots(2,2)

city = 0
for i in range(2): # rows (i) - 0,1
    for j in range(2) : # columns (j) - 0,1
        ax[i][j].plot(days,temperature[city],marker = 'o')
        ax[i][j].set_title(cities[city])
        ax[i][j].set_xlabel("days")
        ax[i][j].set_ylabel("temp")
        ax[i][j].grid(True)
        city = city+1

fig.suptitle("Temperature in cities over the week")
fig.supylabel("Temperature")
fig.supxlabel("Days")
fig.tight_layout()
```

```python
# we can solve it by using single loop instead of nested
# for this we have to flatten the axes
fig,axes = plt.subplots(2,2) # here axes will be a 2d array

axes = axes.flatten() # it will convert it to 1d


for i, ax in enumerate(axes):               # enumerate function idx, val at idx both returns
    ax.plot(days,temperature[i],marker='o')
    ax.set_title(cities[i])
    ax.grid(True)
 
    
fig.suptitle("temperature in Cities over the week")
fig.supylabel("Temperature")
fig.supxlabel("Days")

fig.tight_layout()
```
